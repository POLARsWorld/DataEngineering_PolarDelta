## 1.5. Типы данных и текст
Приведите типы к правильным: даты в `pd.to_datetime`, числа из текста в `int`/`float`. Уберите лишние пробелы в тексте и приведите регистр к нижнему.

**Объяснение:** Какие колонки были неправильного типа? Как чистили текст?

```python
# Ваш код здесь
df['Join Date'] = pd.to_datetime(df['Join Date'], format='mixed', errors='coerce')
df['Last Active Date'] = pd.to_datetime(df['Last Active Date'], format='mixed', errors='coerce')

df['Salary'] = df['Salary'].astype(str).str.replace('AED', '', regex=False).str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.replace('Unknown', '', regex=False).str.strip()

df [['Age', 'Project Count']]   = df[['Age', 'Project Count']].apply(pd.to_numeric, errors='coerce').astype('Int64')
df[['Performance Rating', 'Hours Worked (Week)', 'Engagement Score', 'Salary']] = df[['Performance Rating', 'Hours Worked (Week)', 'Engagement Score', 'Salary']].apply(pd.to_numeric, errors='coerce').astype(float)

str_cols = df.select_dtypes(include=['object', 'str']).columns
df[str_cols] = df[str_cols].apply(lambda s: s.str.strip().str.lower())

df['Gender'] = df['Gender'].astype(str).str.replace({'m': 'male', 'f': 'female', 'other': 'other'})

print(df.dtypes)
```

# Неправильный тип был у: Age, Join Date, Last Active Date, Salary, Gender, Performance Rating, Hours Worked (Week), Engagement Score, Project Count. Для дат смешанный формат ввода иначе None, для int и float смотрел логику столбцов и какие данные в них внесены, если трансформировать нельяза, то None. Для str нахожу все str столбцы (все остальные логические отобрал и уже переделал) и убераю пробелы через strip(), после привожу к нижнему регистру. ## 1.6. Поиск аномалий и выбросов
Иногда в данных есть значения, которые ломают статистику (например, возраст 999 лет, зарплата 0 рублей, отрицательное количество товаров).

1. Выберите одну числовую колонку (например, цену, возраст, количество).
2. Постройте для неё `boxplot` (ящик с усами) или найдите выбросы с помощью правила IQR.
3. Отфильтруйте эти строки (или замените их на `NaN`, если считаете, что они ошибочные).

**Объяснение:** В какой колонке нашли выбросы? Сколько их? Почему решили их удалить (или оставить)?

```python
# Ваш код здесь
import matplotlib.pyplot as plt
import numpy as np

start_na = df['Age'].isna().sum()

df.boxplot('Age', figsize=(4,4))
plt.title(f"Boxplot для {'Age'}")
plt.show()

df['Age'] = df['Age'].abs()

Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.6) #Чтобы возраст был адекватным
IQR = Q3 - Q1

lower_bound = 0
upper_bound = Q3 + 1.5 * IQR

print(f"Q1={Q1}, Q3={Q3}, IQR={IQR}")
print(f"Границы: [{lower_bound}, {upper_bound}]")

outlier_mask = (df['Age'] < lower_bound) | (df['Age'] > upper_bound)
df.loc[outlier_mask, 'Age'] = np.nan

print(f'Количество значений, которые были заменены: {df["Age"].isna().sum() - start_na}')
```

# В колонке возраст 216 выбросов было заменено на None, оставил так как скорее всего это ошибка в конкретном столбце (Люди столько не живут, а другие данные в строке могут быть правдимыми), если выкидывать 20% датасета, то будет большая потеря данных
